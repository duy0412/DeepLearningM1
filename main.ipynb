```python
import pandas as pd
```

```python
df = pd.read_csv("./loan.csv")
df.head()
```

```python
x = df['Experience'].values
y = df['Loan'].values
```

```python
for xi, yi in zip(x, y):
    print(f"{xi} - {yi}") 
```

```python
def f(w, x, y):
    n = len(x)
    loss = 0.0
    for i in range(n):
        y_hat = w[0] + w[1] * x[i]
        loss += (y_hat - y[i])**2
    return loss / n

def f_prime(w, x, y):
    n = len(x)
    dw0 = 0.0
    dw1 = 0.0

    for i in range(n):
        y_hat = w[0] + w[1] * x[i]
        error = y_hat - y[i]

        dw0 += error
        dw1 += error * x[i]

    dw0 = (2/n) * dw0
    dw1 = (2/n) * dw1

    return [dw0, dw1]

def descend(w, alpha, x, y):
    grad = f_prime(w, x, y)
    w0 = w[0] - alpha * grad[0]
    w1 = w[1] - alpha * grad[1]
    return [w0, w1]

def gradient_descend(w, alpha, thred, x, y):
    iter = 0
    while True:
        grad = f_prime(w, x, y)
        loss = f(w, x, y)
        print(f"Iter {iter}: w = {w} \n Loss = {loss}")
        if abs(grad[0]) < thred and abs(grad[1]) < thred:
            break
        w = descend(w, alpha, x, y)
        iter += 1
    return w
```

```python
class LinearRegression:
    def __init__(self, alpha=0.01, thred=1e-6):
        self.alpha = alpha
        self.thred = thred
        self.w = [0.0, 0.0]

    def fit(self, x, y):
        self.w = gradient_descend(self.w, self.alpha, self.thred, x, y)

    def predict(self, x):
        y_hat = []
        for xi in x:
            y_hat.append(self.w[0] + self.w[1] * xi)
        return y_hat
```

```python
model = LinearRegression(alpha=0.01, thred=1e-6)

model.fit(x, y)
```
